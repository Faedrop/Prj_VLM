```python
import os
import csv
import matplotlib.pyplot as plt
from PIL import Image

if os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")

# 1. Afficher le tableau récapitulatif des 500 images
print(f"{'Modèle':<24} | {'Mode':<6} | {'Images':<6} | {'VRAM (Go)':<10} | {'Temps (s)':<10} | {'ROUGE-L':<8}")
print("-" * 76)

with open("results/resume_metriques.csv", "r", encoding="utf-8") as f:
    for row in csv.DictReader(f):
        print(f"{row['modele']:<24} | {row['precision']:<6} | {row['nb_images']:<6} | {row['vram_max_go']:<10} | {row['temps_moyen_sec']:<10} | {row['rouge_l']:<8}")

# 2. Afficher les graphiques générés dans results/
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].imshow(Image.open("results/graphique_vram.png"))
axes[0].axis("off")

axes[1].imshow(Image.open("results/graphique_qualite.png"))
axes[1].axis("off")

plt.tight_layout()
plt.show()

# 3. Afficher la planche d'exemples visuels
plt.figure(figsize=(12, 10))
plt.imshow(Image.open("results/exemples_visuels.png"))
plt.axis("off")
plt.show()
```
